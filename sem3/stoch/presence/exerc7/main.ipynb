```python
from scipy.stats import expon, norm
import matplotlib.pyplot as plt
import numpy as np
```

```python
Z1, Z2 = expon(0.5), norm(2,1)
# a)
x = np.linspace(-2,14)
fZ1 = lambda x: (
    0.5*np.e**(-0.5*x) if x >= 0 else
    0
)
fZ2 = lambda x: 1/(1*np.sqrt(2*np.pi))*np.e**(-1/2*((x-2)/1)**2)
ffZ1 = [fZ1(i) for i in x]
ffZ2 = [fZ2(i) for i in x]
plt.plot(x,ffZ1)
plt.show()
plt.plot(x,ffZ2)
plt.show()

# b)
pb1, pb2 = Z1.cdf(2), Z2.cdf(2)
print(pb1,pb2)
# c)
pc1, pc2 = Z1.ppf(0.90), Z2.ppf(0.90)
print(pc1,pc2)
# d) 95 => 2s
pd2 = (2-2*1,2+2*1)
print(pd2)
```
