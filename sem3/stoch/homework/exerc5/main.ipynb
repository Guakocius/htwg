## 30% keine Vorspeise, 40% keinen Nachtisch, 25% <br>weder " noch "

```python
p_keine_vorspeise = 0.3
p_keinen_nachtisch = 0.4
p_weder_noch = 0.25
p_vorspeise = 1 - p_keine_vorspeise
p_nachtisch = 1 - p_keinen_nachtisch
p_sowohl_alsauch = 1 - p_weder_noch

print(f"P(V) = {p_vorspeise}\t P(~V) = {p_keine_vorspeise}\nP(N) = {p_nachtisch}\t P(~N) = {p_keinen_nachtisch}",
      f"\nP(V,N) = {p_sowohl_alsauch}\t P(~(V,N)) = {p_weder_noch}")
```

```python
p_nN_unter_nV = p_weder_noch / p_keine_vorspeise
p_V_nN = p_keinen_nachtisch - p_weder_noch
p_N_V = p_vorspeise - p_V_nN
p_V_auch_N = p_N_V / p_vorspeise
print(f"a) P(~N | ~V) = {p_nN_unter_nV:.4f}\nb) P(N | V) = {p_V_auch_N:.4f}")
```
