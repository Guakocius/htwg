## Aufgabe 4.1

```python

# a)
x = np.array(list(range(1,6)))
P_x = np.array([0.25, 0.1, 0.1, 0.2, 0.35])

plt.stem(x, P_x)
plt.title("Diagramm der Wahrscheinlichkeitsverteilung")
plt.xticks(x)
plt.show()

# b)
cdf = np.cumsum(P_x)
plt.step(x, cdf)
plt.title("Diagramm der Verteilungsfunktion")
plt.xticks(x)
plt.show()

# c)
def E(x: NDArray, P_x: NDArray):
    X = 0
    for i in range(x.__len__()):
        X += (x[i] * P_x[i])
    return X

e = E(x, P_x)
print(f"E(X): {e}")

# d)
def Var(x: NDArray, P_x: NDArray, e: float):
    sigma_sqr = 0
    for i in range(x.__len__()):
        sigma_sqr += ((x[i] - e)**2 * P_x[i])
    return sigma_sqr

v = Var(x, P_x, e)
print(f"Var(X): {v}")
        
```

## Aufgabe 4.3

### Teilaufgabe 4.3.1 

```python
# a)

# NOTE: valuesX := 1..3, Werte der Zufallsvariable; countsX := count(valuesX_n), gezähltes Vorkommen des Variablenwerts
X_1 = np.array([1, 2, 2, 3, 3, 3, 3, 3])
X_2 = np.array([1, 1, 2, 2, 2, 3, 3, 3])
values1, counts1 = np.unique(X_1, return_counts=True) 
P_X1 = counts1 / len(X_1)
values2, counts2 = np.unique(X_2, return_counts=True)
P_X2 = counts2 / len(X_2)

cdf_X1 = np.cumsum(P_X1)
cdf_X2 = np.cumsum(P_X2)

print(f"CDF(X1): {cdf_X1}\nCDF(X2): {cdf_X2}")

# b)
fig, axs = plt.subplots(2)
axs[0].stem(values1, P_X1)
axs[0].set_title("Wahrscheinlichkeitsverteilung von X1")
axs[0].set_xticks(values1)
axs[1].stem(values2, P_X2)
axs[1].set_title("Wahrscheinlichkeitsverteilung von X2")
axs[1].set_xticks(values2)
plt.tight_layout()

# c)
fig, axs = plt.subplots(2)
axs[0].step(values1, cdf_X1, where="post")
axs[0].set_title("Verteilungsfunktion von X1")
axs[0].set_xticks(values1)
axs[1].step(values2, cdf_X2, where="post")
axs[1].set_title("Verteilungsfunktion von X2")
axs[1].set_xticks(values2)
plt.tight_layout()


```

## Aufgabe 4.8

### Teilaufgabe 4.8.1

```python
# X := Sonnenintensität => 1 | 2 => wenig Sonne | viel Sonne
# Y := Anzahl verkaufter Eiskugeln pro Stunde => 0 | 1 | 2 => keine | einige | viele

#a)
# Prüfen zwecks Gültigkeit der gemeinsamen Wahrscheinlichkeitsverteilung
# I. Alle j für Y[0..n-1] sind nicht negativ
# II. Die Summe aller P(xi, yj) für P(X, Y) muss gleich 1 sein.
P = np.array([[0.1, 0.2, 0.1],[0.1, 0.3, 0.2]]) # Y(j) | Xi => Y(0..2)_X1 | Y(0..2)_X2

n_neg = np.all(P >= 0)
sum = P.sum()
equ_1 = sum == 1

print(f"Sind alle Zahlen nicht negativ? {n_neg}\nIst die Summe aller Elemente gleich 1? {equ_1}")

display(Markdown(f"#### Summa summarum: P(X,Y) ist eine gültige Wahrscheinlichkeitsverteilung => {sum} == 1 && alle i in {P.flatten()} >= 0"))

# b)
P_X = [P[i].sum() for i in range(2)]
P_Y = [P[0][i] + P[1][i] for i in range(3)]

print(f"Randverteilungen:\nP(X = 1): {P_X[0]:.1f}\tP(X = 2): {P_X[1]:.1f}\nP(Y = 0): {P_Y[0]:.1f}\tP(Y = 1): {P_Y[1]:.1f}\tP(Y = 2): {P_Y[2]:.1f}")

# c)
X = np.array([1, 2])
Y = np.array(list(range(3)))
#E_X = X[0] * P_X[0] + X[1] * P_X[1]
E_X = E(X, P_X)
E_Y = E(Y, P_Y)
print(f"E(X): {E_X}")
print(f"E(Y): {E_Y}")

# d)
V_X = Var(X, P_X, E_X)
V_Y = Var(Y, P_Y, E_Y)
print(f"Var(X): {V_X:.2f}")
print(f"Var(Y): {V_Y:.2f}")
```

### Teilaufgabe 4.8.2

```python
# a)
def Cov(X: NDArray, Y: NDArray, EX: float, EY: float):
    return np.sum((X[:, None] - EX) * (Y - EY) * P)
    
cov = Cov(X, Y, E_X, E_Y)
print(f"Cov(X,Y): {cov:.2f}")

# b)
def Corrcoef(cov: float, std_X: float, std_Y: float):
    return cov / (std_X * std_Y)

sigma_X = std_X = np.sqrt(V_X)
sigma_Y = std_Y = np.sqrt(V_Y)

corr = Corrcoef(cov, sigma_X, sigma_Y)
print(f"Corr(X,Y): {corr:.5f}")
```

```python
# d)
# Z := sum(X,Y) => neue Zufallsvariable
Z = np.sum(X[:, None] + Y)
V_Z = Var(X) + V(Y) + 2 * Corrcoef(cov, sigma_X, sigma_Y)
print(f"Z: {Z}")
```
