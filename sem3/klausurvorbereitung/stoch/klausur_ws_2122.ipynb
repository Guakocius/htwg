```python
G_A = [3,2,10,7,0,3,5]
mode = stats.mode(G_A)
print(mode)
mean = np.mean(G_A)
print(mean)
q1 = np.quantile(G_A, 0.25)
q2 = np.quantile(G_A, 0.5)
q3 = np.quantile(G_A, 0.75)
print(q1,q2,q3)
q95 = np.quantile(G_A, 0.95)
print(q95)
std = np.std(G_A, ddof=1)
print(std)

# f)
mu = 1 / 7 * sum([i for i in G_A])
print(mu)

# g)
sigma_sqr = 1 / 7 * sum([(i - mu)**2 for i in G_A])
print(sigma_sqr)
```
