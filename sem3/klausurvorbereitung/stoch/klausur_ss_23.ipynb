### 2.2

```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import combinations

# Daten
V = np.array([62, 47, 15, 37, 53, 56, 93])   # Geschwindigkeit
F = np.array([13, 11, 8, 8, 9, 12, 14])      # Fleischverbrauch
G = np.array([302, 308, 238, 203, 243, 308, 328])  # Größe
A = np.array([37, 55, 55, 42, 50, 50, 55])   # Aggression

data = {
    "G": G,
    "V": V,
    "F": F,
    "A": A
}
import pandas as pd

df = pd.DataFrame(data)

corr = df.corr(method="pearson")
print(corr)

plt.style.use("seaborn-v0_8-whitegrid")

pairs = list(combinations(data.keys(), 2))

fig, axes = plt.subplots(2, 3, figsize=(12, 8))
axes = axes.flatten()

for ax, (x, y) in zip(axes, pairs):
    ax.scatter(data[x], data[y], color="steelblue")
    r = np.corrcoef(data[x], data[y])[0,1]
    ax.set_title(f"{x} vs {y}  (r = {r:.3f})")
    ax.set_xlabel(x)
    ax.set_ylabel(y)

plt.tight_layout()
plt.show()
```
