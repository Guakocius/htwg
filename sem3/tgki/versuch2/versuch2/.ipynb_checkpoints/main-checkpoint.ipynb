## Aufgabe 1

### Zuerst Das Bild einlesen, croppen dessen Graustufen isolieren und deren Mittelwerte und Standardabweichungen berechnen.

```python
img = cv2.imread("./versuch2-1_2.png")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

h, w = gray.shape
step = w // 5

cropped_imgs = [cv2.imread(f"./stufen_corrected/stufe_{i}.png") for i in range(1, 6)]
print("Cropped imgs:", cropped_imgs)
means = [np.mean(i) for i in cropped_imgs]
stds = [np.std(i) for i in cropped_imgs]
print(f"Mittelwerte: {means}\nStandardabweichungen: {stds}")
```

## Aufgabe 2

### Alle Schwarzbilder einlesen und sie in Graubilder übersetzen.

```python
imgs = [cv2.imread(f"./versuch2-2_{i}.png") for i in range(1,11)]
gray = [cv2.cvtColor(i, cv2.COLOR_BGR2GRAY) for i in imgs]
gray = np.float64(gray)
print(gray)
```

### Alle pixelweisen Mittelwerte berechnen und und daraus ein "Bild von Mittelwerten" erstellen.

```python
stack = np.stack(gray, axis=0)
means_px = np.mean(stack.astype(np.float64), axis=0)

cont = means_px.astype(np.float64)
min = cont.min()
max = cont.max()
stretched = (cont - min) / (max - min) * 255

dark = stretched

plt.plot(means_px)
plt.title("Pixelweise Mittelwerte")
plt.show()
plt.plot(stretched)
plt.title("Kontrastmaximierte Mittelwerte")
plt.show()
```

### Liest Dunkelbild ein, subtahiert von einem Eingabebild und gibt das korrigierte Bild aus.

```python
# img: Eingabebild, src: Quellpfad des Dunkelbildes
def sub_from_img(src, img):
    d_src = cv2.imread(src)
    img = np.float64(img)
    mean = np.mean(img)
    means_mean = np.mean(mean)
    #means_norm = means / means_mean # Mittelwert Kontrastmaximierung := 1
    
    return img - d_src
```

## Aufgabe 3

### 10 Weißbilder einlesen, ihren Mittelwert berechnen, und davon das Dunkelbild berechnen.

```python
def stretch_255(img):
    mi, ma = img.min(), img.max()
    stretched = (img - mi) / (ma - mi) * 255
    return stretched.astype(np.uint8)


dark_files = sorted(Path(".").glob("schwarz*.png"))
print("Gefundene Darkframes:", dark_files)

dark_imgs = [
    cv2.imread(str(f), cv2.IMREAD_GRAYSCALE).astype(np.float64)
    for f in dark_files
]

dark = np.mean(np.stack(dark_imgs, axis=0), axis=0)
print("Darkframe shape:", dark.shape)

cv2.imwrite("dark_mean.png", dark.astype(np.uint8))


white_imgs = [
    cv2.imread(f"versuch2-3_{i}.png", cv2.IMREAD_GRAYSCALE).astype(np.float64)
    for i in range(1, 11)
]

white = np.mean(np.stack(white_imgs, axis=0), axis=0)
print("White shape:", white.shape)

white_corr = white - dark
white_corr = np.clip(white_corr, 0, None)


white_stretched = stretch_255(white_corr)
cv2.imwrite("weissbild_kontrastmaximiert.png", white_stretched)

plt.imshow(white_stretched, cmap='gray')
plt.title("Weißbild kontrastmaximiert")
plt.colorbar()
plt.show()


for f in sorted(Path(".").glob("schwarz*.png")):
    img = cv2.imread(str(f), cv2.IMREAD_GRAYSCALE)
    print(f.name, img.shape)



white_mean = np.mean(white_corr)
white_norm = white_corr / white_mean
print("White normalized mean:", white_norm.mean())  # sollte ≈ 1 sein

input_img = cv2.imread("versuch2-1.png", cv2.IMREAD_GRAYSCALE).astype(np.float64)
input_img_resized = cv2.resize(input_img, (640, 480), interpolation=cv2.INTER_CUBIC)

input_corr = input_img_resized - dark
input_corr = np.clip(input_corr, 0, None)

flatfield = input_corr / white_norm

flatfield_stretched = stretch_255(flatfield)
cv2.imwrite("versuch2-3_dividiert.png", flatfield_stretched)

plt.imshow(flatfield_stretched, cmap="gray")
plt.title("Flatfield-korrigiertes Bild (dividiert)")
plt.colorbar()
plt.show()
```
