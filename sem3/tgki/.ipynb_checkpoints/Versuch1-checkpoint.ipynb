```python
 for i in range(10, 71, 3):
    df = pd.read_csv(f"./Versuch1-{i}.csv", sep=";", decimal=",", skiprows=1)
    mean = np.mean(df["(V)"])
    print(f"\u001b[3;35mMean for {i}cm: {mean}\u001b[0m")
    std = np.std(df["(V)"], ddof=1)
    print(f"\u001b[1;34mStd for {i}cm: {std}\u001b[0m\n")
```
